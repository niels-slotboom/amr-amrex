```python
import yt
import numpy as np
from IPython.display import Image, display

ds = yt.load("../raw/test")

# 1. Define the derived field function
def _sin_phi(field, data):
    return np.sin(10.0*data[("boxlib", "phi")])

# 2. Add the field to the dataset
ds.add_field(
    name=("boxlib", "sin_phi"),
    function=_sin_phi,
    sampling_type="cell",
    units="",  # specify units if applicable, e.g., "1/cm" or "" for dimensionless
)

# 3. Plot the new derived field
slc = yt.SlicePlot(ds, "z", ("boxlib", "phi"))
slc.set_cmap(("boxlib", "phi"), "berlin")  # Set colormap to berlin
slc.annotate_grids(periodic=False)

# 1. Set pixel grid buffer resolution (default is 800x800)
slc.set_buff_size((2000, 2000))

# 2. Adjust figure canvas size in inches (optional)
slc.set_figure_size(10)  # Width in inches

# 3. Increase output DPI when saving
slc.save("phi_slice_hd.png", mpl_kwargs={"dpi": 600})
display(Image("phi_slice_hd.png"))
```
